1. AI 模型中的權重通常使用什麼型別儲存？FP32

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int | 8/16/32/64 | x | 計數、索引、量化後的 AI 模型 |

3. 哪種表示方式最常見於 AI 訓練？FP16 / BF16
4. 哪種表示方式最常見於 AI 推論（Inference）？FP16 / BF16

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。正確性/效率
